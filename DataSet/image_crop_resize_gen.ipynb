```python
from PIL import Image
import os
import numpy as np

```

```python
it = 0
```

```python
def split_and_resize(image_path, x_size, y_size, resize_factor, subimages_folder, resized_folder):
  global it
  """
  This function splits an image into non-overlapping subimages of size (x_size, y_size)
  and resizes them to a factor of 1/resize_factor.

  Args:
      image_path: Path to the image file.
      x_size: Width of the subimage.
      y_size: Height of the subimage.
      resize_factor: Factor by which to reduce the resolution (e.g., 2 for half size).

  """
  # Open the image
  image = Image.open(image_path)
  image_width, image_height = image.size

  # Calculate the maximum number of rows and columns
  max_rows = image_height // y_size
  max_cols = image_width // x_size

  subimages = []
  for row in range(max_rows):
    for col in range(max_cols):
      # Define coordinates for the subimage
      start_x = col * x_size
      start_y = row * y_size
      end_x = min(start_x + x_size, image_width)
      end_y = min(start_y + y_size, image_height)

      subimage = image.crop((start_x, start_y, end_x, end_y))
      subimage.save(os.path.join(subimages_folder,str(it)+'.png'))
      width, height = subimage.size
      new_width = int(width / resize_factor)
      new_height = int(height / resize_factor)
      resized_subimage = subimage.resize((new_width, new_height), resample=Image.BICUBIC)
      resized_subimage.save(os.path.join(resized_folder, str(it)+'.png'))
      it += 1

def split_and_resize_all_images(root_folder, x_size, y_size, resize_factor, subimages_folder, resized_folder):
    """
    Splits and resizes all images within a folder and its subfolders,
    saving subimages and resized images in separate folders with the same name.

    Args:
        root_folder (str): Path to the root folder containing images.
        x_size (int): Width of the subimage.
        y_size (int): Height of the subimage.
        resize_factor (int): Factor by which to reduce the resolution.
    """

    for root, _, files in os.walk(root_folder):
        for file in files:
            if file.lower().endswith(('.jpg', '.jpeg', '.png')):
                image_path = os.path.join(root, file)

                split_and_resize(image_path, x_size, y_size, resize_factor, subimages_folder, resized_folder)
```

```python
def grayscale_to_rgb_numpy(grayscale_image):
  """
  Converts a grayscale image to RGB using vectorized operations (NumPy).

  Args:
      grayscale_image (np.ndarray): Grayscale image as a NumPy array.

  Returns:
      np.ndarray: RGB image as a NumPy array.
  """
  return np.dstack([grayscale_image] * 3)

def converter(root_folder):
  for root, _, files in os.walk(root_folder):
      for file in files:
          if file.lower().endswith(('.jpg', '.jpeg', '.png')):
              image_path = os.path.join(root, file)
              arr = np.array(Image.open(image_path))
              if len(arr.shape) == 3:
                arr = arr[:,:,0]
              Image.fromarray(grayscale_to_rgb_numpy(arr)).save(image_path)
```

```python
converter = np.vectorize(converter)
converter(r'G:\My Drive\Moon Mapping\DataSet\Test')
```
